# **Q1**

```python
roll_no = input("Enter your roll number: ")
L = [int(digit) * 10 for digit in roll_no]

print("Original List:", L)
```

```python
L.append(50)
print("After append(50):", L)

L.insert(2, 70)
print("After insert(2, 70):", L)
```

```python
L.remove(50)
print("After remove(50):", L)

L.pop(2)
print("After pop(2):", L)
```

```python
L.sort()
print("Ascending Order:", L)

L.sort(reverse=True)
print("Descending Order:", L)
```

```python
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])
```

```python
average = sum(L) / len(L)
greater_than_average = [x for x in L if x > average]

print("Average:", average)
print("Elements greater than average:", greater_than_average)
```

# **Q2**

```python
# assuming L created in Q1
scores = tuple(L[:8])
print("Scores Tuple:", scores)
```

```python
highest = max(scores)
highest_index = scores.index(highest)

lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest Score:", highest)
print("Index of Highest Score:", highest_index)
print("Lowest Score:", lowest)
print("Lowest Score Appears:", lowest_count, "time(s)")
```

```python
reversed_scores = list(scores[::-1])
print("Reversed Tuple as List:", reversed_scores)
```

```python
search_score = int(input("Enter a score to search: "))

if search_score in scores:
    print(f"First occurrence index of {search_score}: {scores.index(search_score)}")
else:
    print(f"{search_score} is not present in the tuple.")
```

```python
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)

print("Reason: Tuples do not allow item assignment, but lists do.")
```

```python
first_score, second_score, *remaining_scores = scores

print("First Score:", first_score)
print("Second Score:", second_score)
print("Remaining Scores:", remaining_scores)
print("Org Scores:", scores)
```

# **Q3**

```python
import random
```

```python
roll_number = int(input("Enter your roll number: "))
random.seed(roll_number)
numbers = [random.randint(100, 900) for _ in range(100)]
print("Random List:")
print(numbers)
```

```python
odd_numbers = [num for num in numbers if num % 2 != 0]

print("\nOdd Numbers:")
print(odd_numbers)
print("Count of Odd Numbers:", len(odd_numbers))
```

```python
even_numbers = [num for num in numbers if num % 2 == 0]

print("\nEven Numbers:")
print(even_numbers)
print("Count of Even Numbers:", len(even_numbers))
```

```python
def is_prime(num):
    if num < 2:
        return False

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False

    return True

prime_numbers = [num for num in numbers if is_prime(num)]

print("\nPrime Numbers:")
print(prime_numbers)
print("Count of Prime Numbers:", len(prime_numbers))
```

```python
frequency = {}

for num in numbers:
    frequency[num] = frequency.get(num, 0) + 1

most_frequent_number = max(frequency, key=frequency.get)
most_frequent_count = frequency[most_frequent_number]

print("\nMost Frequently Occurring Number:", most_frequent_number)
print("Number of Occurrences:", most_frequent_count)
```

# **Q4**

```python
roll_number = input("Enter your 8-digit roll number: ")
digits = []

for digit in roll_number:
    digits.append(int(digit))

A = set()

for digit in digits:
    A.add(digit * 7)

B = set()

for digit in digits:
    B.add(digit * 9)

print("Set A:", A)
print("Set B:", B)
```

```python
print("Union of A and B:", A.union(B))
```

```python
print("Intersection of A and B:", A.intersection(B))
```

```python
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
```

```python
print("Symmetric Difference:", A.symmetric_difference(B))
```

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

```python
x = int(input("Enter a value X to remove from Set A: "))

A.discard(x)

print("Set A after discard:", A)
```

# **Q5**

```python
my_dict = {
    "name": "Raghav",
    "roll_no": "12345678",
    "branch": "CSE",
    "age": 18,
    "city": "Delhi"
}

print("Original Dictionary:", my_dict)
```

```python
my_dict["location"] = my_dict.pop("city")
print("\nAfter renaming city to location:", my_dict)
```

```python
my_dict["cgpa"] = 9.0
print("\nAfter adding CGPA:", my_dict)
```

```python
my_dict["age"] = my_dict["age"] + 1
print("\nAfter increasing age:", my_dict)
```

```python
dict_pop = my_dict.copy()
dict_del = my_dict.copy()
removed_branch = dict_pop.pop("branch")
print("\nDictionary after using pop():", dict_pop)
print("Removed value:", removed_branch)

del dict_del["branch"]
print("Dictionary after using del:", dict_del)

# NOTE: pop() removes the key and returns its value, while del only removes the key
# and does not return the deleted value.
```

```python
print("\nKey-Value Pairs:")

for key, value in my_dict.items():
    print(key, "→", value)
```

```python
if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail key is not present in the dictionary.")
```

```python
friend_dict = {
    "name": "Aman",
    "roll_no": "87654321",
    "branch": "ECE",
    "age": 19,
    "city": "Mumbai"
}

merged_dict = {**my_dict, **friend_dict}

print("\nFriend Dictionary:", friend_dict)
print("Merged Dictionary:", merged_dict)
```

```python
string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nDictionary containing only string values:", string_values)
```
